```python
%%writefile atm_simulation.cpp

#include <iomanip>
#include <iostream>
#include <limits>

using namespace std;

class ATM {
private:
    double balance;

public:
    ATM(double initialBalance = 1000.0) {
        balance = initialBalance;
    }

    double checkBalance() const {
        return balance;
    }

    bool deposit(double amount) {
        if (amount <= 0) {
            return false;
        }

        balance += amount;
        return true;
    }

    bool withdraw(double amount) {
        if (amount <= 0 || amount > balance) {
            return false;
        }

        balance -= amount;
        return true;
    }
};

class ATMController {
private:
    ATM& atm;

    void clearInvalidInput() {
        cin.clear();
        cin.ignore(numeric_limits<streamsize>::max(), '\n');
    }

    bool readAmount(const string& prompt, double& amount) {
        cout << prompt;

        if (!(cin >> amount)) {
            cout << "Invalid amount. Please enter a number." << endl;
            clearInvalidInput();
            return false;
        }

        return true;
    }

public:
    ATMController(ATM& account) : atm(account) {}

    void displayMenu() {
        cout << "\n======================================" << endl;
        cout << "          ATM SIMULATION" << endl;
        cout << "======================================" << endl;
        cout << "1. Check Balance" << endl;
        cout << "2. Deposit Money" << endl;
        cout << "3. Withdraw Money" << endl;
        cout << "4. Exit" << endl;
    }

    void run() {
        int choice;

        cout << fixed << setprecision(2);
        cout << "Welcome to the ATM!" << endl;

        while (true) {
            displayMenu();
            cout << "Choose an option (1-4): ";

            if (!(cin >> choice)) {
                cout << "Invalid option. Enter a number from 1 to 4." << endl;
                clearInvalidInput();
                continue;
            }

            if (choice == 1) {
                cout << "Current balance: Rs. "
                     << atm.checkBalance() << endl;
            }

            else if (choice == 2) {
                double amount;

                if (!readAmount(
                        "Enter amount to deposit: Rs. ",
                        amount
                    )) {
                    continue;
                }

                if (atm.deposit(amount)) {
                    cout << "Deposit successful." << endl;
                    cout << "New balance: Rs. "
                         << atm.checkBalance() << endl;
                } else {
                    cout << "Amount must be greater than zero."
                         << endl;
                }
            }

            else if (choice == 3) {
                double amount;

                if (!readAmount(
                        "Enter amount to withdraw: Rs. ",
                        amount
                    )) {
                    continue;
                }

                if (amount <= 0) {
                    cout << "Amount must be greater than zero."
                         << endl;
                }

                else if (amount > atm.checkBalance()) {
                    cout << "Insufficient balance." << endl;
                }

                else if (atm.withdraw(amount)) {
                    cout << "Withdrawal successful." << endl;
                    cout << "Remaining balance: Rs. "
                         << atm.checkBalance() << endl;
                }
            }

            else if (choice == 4) {
                cout << "Thank you for using the ATM. Goodbye!"
                     << endl;
                break;
            }

            else {
                cout << "Invalid option. Select a number from 1 to 4."
                     << endl;
            }
        }
    }
};

int main() {
    ATM account(1000.0);
    ATMController controller(account);

    controller.run();

    return 0;
}
```
